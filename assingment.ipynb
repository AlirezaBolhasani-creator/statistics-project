### PROBLEM 6 : Convolution 

```python
import numpy as np
import matplotlib.pyplot as plt

# A:
#define function in order to calculate convolution:
def convolution(x_samples, y_samples):
    output_length = len(x_samples) + len(y_samples) - 1
    result = [0] * output_length
    for t in range(output_length):
        for k in range(len(y_samples)):
            if 0 <= t - k < len(x_samples):  
                result[t] += x_samples[t - k] * y_samples[k]
    
    return result

# B:
#define pdf of x:
x_vals = np.arange(-5, 6, 1)
pdf_x = np.zeros_like(x_vals, dtype=float)
pdf_x[x_vals == 0] = 0.2
pdf_x[x_vals == -1] = 0.16
pdf_x[x_vals == 1] = 0.16
pdf_x[x_vals == -2] = 0.12
pdf_x[x_vals == 2] = 0.12
pdf_x[x_vals == -3] = 0.08
pdf_x[x_vals == 3] = 0.08
pdf_x[x_vals == -4] = 0.04
pdf_x[x_vals == 4] = 0.04
pdf_x[x_vals == -5] = 0
pdf_x[x_vals == 5] = 0
#define pdf of y:
y_vals = np.arange(-5, 6, 1)
pdf_y = np.zeros_like(y_vals, dtype=float)
pdf_y[y_vals == 0] = 0.2
pdf_y[y_vals == -1] = 0.1
pdf_y[y_vals == 1] = 0.1
pdf_y[y_vals == -2] = 0.05
pdf_y[y_vals == 2] = 0.05
pdf_y[y_vals == -3] = 0.1
pdf_y[y_vals == 3] = 0.1
pdf_y[y_vals == -4] = 0.15
pdf_y[y_vals == 4] = 0.15
pdf_y[y_vals == -5] = 0
pdf_y[y_vals == 5] = 0
#calculate convoulution pdf:
pdf_x_plus_y = convolution(pdf_x, pdf_y)
#plot PDFs and convolution
plt.figure()
plt.stem(x_vals, pdf_x)
plt.title('PDF of X')
plt.xlabel('X')
plt.ylabel('Probability')
plt.show()

plt.figure()
plt.stem(y_vals , pdf_y)
plt.title('PDF of Y')
plt.xlabel('Y')
plt.ylabel('Probability')
plt.show()

x_plus_y_vals = np.arange(y_vals[0] + x_vals[0], y_vals[-1] + x_vals[-1] + 1, 1)
plt.figure(figsize=(6, 6))
plt.stem(x_plus_y_vals, pdf_x_plus_y)
plt.title('PDF of X + Y')
plt.xlabel('X + Y')
plt.ylabel('Probability')
plt.show()
```

```python
#importing libraries:
from functions import p8_pdf
import matplotlib.pyplot as plt
import numpy as np
from scipy.fft import fft, ifft
from scipy.stats import gaussian_kde 

# A: getting samples and show them:
samples = []
for i in range(10000):
    z = p8_pdf()
    samples.append(z)
plt.hist(samples, bins = 100, linewidth = 0.1)
plt.title("10000 samples of p8")
plt.xlabel("X")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()
# B: moments from 1 to 10
moments = [np.mean(np.array(samples)**i) for i in range(1, 11)]
for i, moment in enumerate(moments, 1):
    print(f"Moment {i}: {moment}")
# Plot the moments
plt.bar(range(1, 11), moments)
plt.title('First 10 Moments of the Distribution (Part ب)')
plt.xlabel('Moment Number')
plt.ylabel('Moment Value')
plt.show()
    
# C: Recreate and plot PDF from 3, 5, and 10 moments
def reconstruct_pdf(moments, num_points=1000):
    x = np.linspace(-5, 5, num_points)
    reconstructed_pdf = np.zeros_like(x, dtype=complex)
    
    for n, moment in enumerate(moments, start=1):
        reconstructed_pdf += moment * np.exp(-1j * 2 * np.pi * n * x / len(x))  # Corrected frequency scaling
    
    reconstructed_pdf = np.real(ifft(reconstructed_pdf))
    reconstructed_pdf = np.fft.fftshift(reconstructed_pdf)  # Shift zero frequency to center
    x = np.fft.fftshift(x)  # Shift x-axis accordingly
    
    return x, reconstructed_pdf

for N in [3, 5, 10]:
    x, reconstructed_n = reconstruct_pdf(moments[:N])
    reconstructed_n = np.maximum(reconstructed_n,0)
    all = sum(reconstructed_n)
    if all > 0 :
        reconstructed_n /= all
    plt.plot(x, reconstructed_n, label=f"Reconstruction ({N} moments)")
   
kde = gaussian_kde(samples, bw_method=0.2) 
x_vals = np.linspace(min(samples), max(samples), 1000)
kde_vals = kde(x_vals)
plt.plot(x_vals, kde_vals, label='KDE(orginal pdf)',linestyle='--', color='black')

plt.hist(samples, bins=50, density=True, alpha=0.2, color='green', label='Original Histogram')
plt.title(f"PDF Reconstruction Using {N} Moments")
plt.xlabel('Values')
plt.ylabel('Density')
plt.legend()
plt.show()
```

### PROBLEM 9: Inequalities 

```python
#importing libraries:
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
from scipy.stats import poisson

# A:
#Standard gaussian:
miu = 0
sigma = 1
N = 10000
samples = np.random.normal(miu, sigma, N)
k = np.linspace(-5, 5, 100)
cheb_probs = []
theoretical_bound = []
new_k = []
for ki in k:
    if ki >= 0.95:
        prob = np.mean(np.abs(samples - miu) > ki * sigma)
        cheb_probs.append(prob)
        theoretical_bound.append(1 / ki**2)
        new_k.append(ki)
#plot 1/k^2 vs p(|x -miu|)
plt.plot(new_k, theoretical_bound, label="1 / k²", color='black')
plt.plot(new_k, cheb_probs, label="Empirical P(|X - μ| > kσ)", color='blue')
plt.title("Chebyshev Inequality Visualization")
plt.xlabel("k")
plt.ylabel("Probability")
plt.grid(True)
plt.legend()
plt.show()

# B:
#poisson with lambda = 5
lam = 5
N = 10000
data = poisson.rvs(mu=lam, size=N)
y = np.arange(0, max(data)+1)
cdf_poisson = poisson.cdf(y, lam)
miu = np.mean(data)
k = np.arange(1, max(data)+1)  # Avoiding division by zero for miu / k
miu_per_k = [miu / ki for ki in k]
#plot miu / k vs poisson CDF :
plt.plot(k,1 - cdf_poisson[:len(k)], label=f'CDF (λ={lam})', color='blue')
plt.plot(k, miu_per_k, label="miu / k", color="red")
plt.title('miu / k vs Poisson CDF')
plt.xlabel('k (Poisson values)')
plt.ylabel('Probability / miu / k')
plt.legend()
plt.grid(True)
plt.show()
```

### PROBLEM 10 : A Simple Example of Bayesian Statistics 

```python
#importing libraries:
import numpy as np
import matplotlib.pyplot as plt

# A:
#define function in order to calculate pdf:
def posterior_pdf(p, heads=60, tails=40):
    return (p ** heads) * ((1 - p) ** tails)
p_values = np.linspace(0, 1, 1000)
posterior_values = posterior_pdf(p_values)
#plot first pdf
plt.plot(p_values, posterior_values, label='Posterior PDF')
plt.xlabel('P (Probability of Heads)')
plt.ylabel('Density')
plt.title('PDF after 60 Heads and 40 Tails')
plt.legend()
plt.grid(True)
plt.show()

# B:
#define function in order to calculate updated pdf:
def updated_posterior_pdf(p, heads1=60, tails1=40, heads2=30, tails2=70):
    return (p ** heads1) * ((1 - p) ** tails1) * (p ** heads2) * ((1 - p) ** tails2)

updated_posterior_values = updated_posterior_pdf(p_values)
#plot second pdf 
plt.plot(p_values, updated_posterior_values, label='Updated Posterior PDF', color='orange')
plt.xlabel('P (Probability of Heads)')
plt.ylabel('Density')
plt.title('Updated PDF after 30 Heads and 70 Tails')
plt.legend()
plt.grid(True)
plt.show()

# C:
print("It means we had no knowledge about the probability, so we assumed all values were equally possible and for updates our work is easier")
```

```python
#importing libraries:
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, multivariate_normal

# A: Load the data
df = pd.read_csv('data.csv')
print("columns:")
print(df.columns.tolist())
print(f"number of rows: {len(df)}")

# B: Remove rows without price
df_clean = df.dropna(subset=['price'])

# C: Save the cleaned data
df_clean.to_csv('data-clean.csv', index=False)
print("new data set is saved: data-clean.csv")

# D: Histogram and Gaussian Fit
plt.figure(figsize=(10, 5))
plt.hist(df_clean['price'], bins=50, density=True, color='skyblue', alpha=0.7, label='datas')

# Fit Gaussian
mu, std = norm.fit(df_clean['price'])
xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 100)
p = norm.pdf(x, mu, std)
plt.plot(x, p, 'r', linewidth=2, label=f'normal distribution\nμ={mu:.0f}, σ={std:.0f}')
plt.title("Histogram for House Prices")
plt.xlabel('Price')
plt.ylabel('Density')
plt.legend()
plt.grid(True)
plt.show()

# E: Calculate area using price and price_per_size (before plotting)
df_clean['price'] = pd.to_numeric(df_clean['price'], errors='coerce')
df_clean['price_per_size'] = pd.to_numeric(df_clean['price_per_size'], errors='coerce')

df_clean = df_clean[df_clean['price_per_size'].notna() & (df_clean['price_per_size'] != 0)]
df_clean['area'] = df_clean['price'] / df_clean['price_per_size']
print("Recalculating areas using price and price_per_size.")

plt.figure(figsize=(8, 6))
plt.scatter(df_clean['area'], df_clean['price'], alpha=0.6, color='blue', edgecolor='k')
plt.title("Area vs Price")
plt.xlabel('Area')
plt.ylabel('Price')
plt.grid(True)
plt.show()

# F: Fit 2D Gaussian to Price and Area
data_2d = df_clean[['price', 'area']].dropna()
mean_vec = data_2d.mean().values
cov_matrix = np.cov(data_2d.T)

print("2D mean:", mean_vec)
print("Covariance matrix:\n", cov_matrix)

# Plotting 2D Gaussian:
x = np.linspace(data_2d['price'].min(), data_2d['price'].max(), 100)
y = np.linspace(data_2d['area'].min(), data_2d['area'].max(), 100)
X, Y = np.meshgrid(x, y)
pos = np.dstack((X, Y))
rv = multivariate_normal(mean_vec, cov_matrix)
Z = rv.pdf(pos)

plt.figure(figsize=(8, 6))
plt.contourf(X, Y, Z, cmap='viridis', levels=30)
plt.scatter(data_2d['price'], data_2d['area'], s=10, alpha=0.5, color='white')
plt.title('2D Gaussian Fit for Price and Area')
plt.xlabel('Price')
plt.ylabel('Area')
plt.colorbar(label='Probability Density')
plt.show()
```
